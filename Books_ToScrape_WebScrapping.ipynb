```python
import requests
from bs4 import BeautifulSoup
```

```python
base_url = "https://books.toscrape.com/catalogue/page-{}.html"
```

```python
all_books = []

for page in range(1, 51):  # site has 50 pages
    print(f"Scraping page {page}...")
    
    url = base_url.format(page)
    response = requests.get(url)
    
    if response.status_code != 200:
        break  # stop when pages end

    soup = BeautifulSoup(response.text, "html.parser")

    books = soup.find_all("article", class_="product_pod")

    for book in books:
        title = book.h3.a["title"]
        price = book.find("p", class_="price_color").text
        rating = book.p["class"][1]
        all_books.append([title, price, rating])

print(f"\nTotal books scraped: {len(all_books)}")
for b in all_books[:10]:
    print(b)
```

```python
import pandas as pd
df = pd.DataFrame(all_books, columns=["Title", "Price", "Rating"])
```
